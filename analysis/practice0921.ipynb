# 2026년 상반기 인천공항에서 화물 전용 항공편은 전체 운항에서 얼마나 차지하며, 환적화물 운송에는 얼마나 기여하는가?

- 전체 운항 구조 파악


```python
total_flights = df['운항(편)'].sum()
total_flights
```

```python
cargo_flights = df[df['여객_화물'] == '화물']['운항(편)'].sum()

cargo_flights
```

```python
cargo_ratio = (cargo_flights/total_flights) * 100

print(f'전체 운항편 중 화물 운항편 비율: {cargo_ratio:.2f}%')
```

- 2026년 1월부터 6월까지 전체 운항편 중 화물 전용 운항편은 11.57%이다.

> 이 11.57%의 화물 전용 운항편은 어떤 항공사로 구성되어 있을까?
 - 화물 전용편을 운항한 항공사 목록은?
  - 그 중 화물 운항편이 많은 항공사는 어디?

```python
filtered  = df[df['여객_화물'] == '화물']
```

```python
filtered['여객_화물'].unique()
```

```python
cargo_by_airline = filtered.groupby('항공사명')['운항(편)'].sum().sort_values(ascending= False)

cargo_by_airline
```

화물 전용편은 전체 운항의 11.57%에 불과하지만, 전체 환적화물 운송에서는 얼마나 기여하고 있을까?

- 화물 전용기의 환적화물 기여율
- 여객기와 화물기의 편당 환적화물량 비교
- 항공사별 화물 운항편 수 순위(자주 운항한), 환적화물량 순위(많이 운송한) 확인

```python
cargo_transfer = filtered['환적화물(kg)'].sum()
total_transfer = df['환적화물(kg)'].sum()
```

```python
transfer_ratio = ((cargo_transfer/total_transfer)* 100)

print(f'전체 환적화물 중 화물 전용기의 기여율: {transfer_ratio:.2f}%')
```

```python
cargo_per_flight = cargo_transfer / cargo_flights
cargo_per_flight
```

```python
print(f'화물기편당환적화물량: {cargo_per_flight:.2f}kg/편')
```

> 대략 화물기 한 편당 16,248kg의 환적화물을 운송했다..
